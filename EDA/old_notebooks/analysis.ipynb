```python
import pandas as pd
import matplotlib.pyplot as plt

# 載入 CSV 資料
csv_path = './raw_data/male_data.csv'   # 改成 female_data.csv 來讀取女性資料
df = pd.read_csv(csv_path)

# 查看資料的基本描述
print(df.describe())

# 繪製散佈圖範例：以 age 為 x 軸，cholesterol 為 y 軸，依據 Second_Stroke 作顏色區分
plt.figure(figsize=(8, 6))
scatter = plt.scatter(df['age'], df['cholesterol'], c=df['Second_Stroke'],
                      cmap='coolwarm', alpha=0.7)  # cmap 可依喜好調整
plt.xlabel('Age')
plt.ylabel('Cholesterol')
plt.title('Scatter Plot of Age vs Cholesterol\n(Colored by Second_Stroke)')
plt.colorbar(scatter, label='Second_Stroke')
plt.show()
```

```python
pd.plotting.scatter_matrix(df[['age', 'cholesterol', 'LDL', 'HbA1c']],
                           figsize=(10, 10),
                           diagonal='kde',
                           color=df['Second_Stroke'].map({0: 'blue', 1: 'red'}))
plt.suptitle('Scatter Matrix of Selected Features')
plt.show()
```
